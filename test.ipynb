```python
from psychopy.iohub import launchHubServer
from psychopy import visual
import pprint # 用于漂亮地打印字典
```

```python
# 1. 建立一个最简单的连接 (用 Dummy 模式即可，或者连 Eyelink)
# 这里演示用 Dummy，方便你直接复制运行
win = visual.Window([800, 600])
```

```python
io = launchHubServer(
    window=win, 
    # 使用 Eyelink 路径但没连机器可能会卡住，这里为了查看参数，用 Mouse 模拟路径
    # 如果你连了机器，换回 sr_research 的路径
    **{'eyetracker.hw.mouse.EyeTracker': {'name': 'tracker'}}
)
```

```python
# 2. 获取设备
tracker = io.devices.tracker
```

```python
print(tracker)
```

```python
# 3. 【核心】打印出 iohub 正在使用的完整配置
print("\n" + "="*40)
print("  当前眼动仪的完整配置 (含默认值)")
print("="*40)
pprint.pprint(tracker.configuration)  # 这就是你要找的所有参数！
print("="*40 + "\n")

win.close()
io.quit()
```
